# YOLOv8 → ONNX export для Poker Terminal
Запусти все ячейки по порядку (Runtime → Run all), потом скачай `playing_cards.onnx`

```python
# 1. Устанавливаем ultralytics
!pip install ultralytics -q
```

```python
# 2. Скачиваем предобученные веса
import urllib.request
url = 'https://huggingface.co/mustafakemal0146/playing-cards-yolov8/resolve/main/playing_cards_model_0_playing-cards-colab.pt'
urllib.request.urlretrieve(url, 'playing_cards.pt')
print('Downloaded!')
```

```python
# 3. Проверяем классы модели (должно быть 52: Ah, Kh, ... 2c)
from ultralytics import YOLO
model = YOLO('playing_cards.pt')
print('Classes:', model.names)
```

```python
# 4. Экспортируем в ONNX (imgsz=640, opset=12 — совместим с onnxruntime-web)
model.export(format='onnx', imgsz=640, opset=12, simplify=True)
print('Export done: playing_cards.onnx')
```

```python
# 5. Скачиваем файл из Colab
from google.colab import files
files.download('playing_cards.onnx')
```
